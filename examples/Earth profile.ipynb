```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams
import matplotlib.font_manager
rcParams['font.family'] = 'serif'
rcParams['font.serif'] = ['Computer Modern']
```

# Import PREM data

```python

PREM = np.loadtxt("../src/data/PREM.rho.dat")
```

```python
fig, axs = plt.subplots(1,1, figsize=(10*1.618,10))

axs.plot(PREM[:,0]/1.e5,PREM[:,1],'C0',linewidth=2)

axs.set_xlabel("Radius [km]",fontsize=30)
axs.set_ylabel(r"$\rho\;{\rm [g/cm^3]} $",fontsize=30)

#axs.minorticks_on()
axs.tick_params(direction='in',length=7,width=2,top=True,right=True,labelsize=22)

axs.spines['bottom'].set_linewidth(2)
axs.spines['top'].set_linewidth(2)
axs.spines['left'].set_linewidth(2)
axs.spines['right'].set_linewidth(2)

axs.set_ylim(bottom=0)

fig.tight_layout()

plt.savefig("PREM.png", format='png')
plt.savefig("PREM.pdf", format='pdf')
plt.show()
```
