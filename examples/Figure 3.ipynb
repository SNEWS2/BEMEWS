```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import rcParams
import matplotlib.font_manager
rcParams['font.family'] = 'serif'
rcParams['font.serif'] = ['Computer Modern']
```

# Import data

```python

Pvsl= np.loadtxt("BEMEWS:PREM:-40,124.306:7.42e-05,0.0024428,33.44,8.57,49.2,197:296,1,60:PvsE:lambda=8.18654e+08cm.dat",skiprows=1)
```

```python
fig, axs = plt.subplots(2,1, figsize=(5*1.618,10), sharex=True)

axs[0].plot(Pvsl[:,0],Pvsl[:,19],'C0',linewidth=2,label = r"$P_{e1}$")
axs[0].plot(Pvsl[:,0],Pvsl[:,20],'C1',linewidth=2,label = r"$P_{e2}$")

axs[1].plot(Pvsl[:,0],Pvsl[:,28],'C0',linewidth=2,label = r"${\bar P}_{e1}$")
axs[1].plot(Pvsl[:,0],Pvsl[:,29],'C1',linewidth=2,label = r"${\bar P}_{e2}$")

axs[1].set_xlabel("Neutrino Energy [MeV]",fontsize=30)

axs[0].set_ylabel(r"$P_{e1}, P_{e2} $",fontsize=30)
axs[1].set_ylabel(r"${\bar P}_{e1}, {\bar P}_{e2} $",fontsize=30)

for i in range(2):
    axs[i].tick_params(direction='in',length=7,width=2,top=True,right=True,labelsize=22)
    axs[i].legend(fontsize=22)

    axs[i].spines['bottom'].set_linewidth(2)
    axs[i].spines['top'].set_linewidth(2)
    axs[i].spines['left'].set_linewidth(2)
    axs[i].spines['right'].set_linewidth(2)

    axs[i].set_ylim(bottom=0,top=1)

fig.tight_layout()

plt.savefig("PvsE.-40.png", format='png')
plt.savefig("PvsE.-40.pdf", format='pdf')
plt.show()
```
